# The seminar
* seminar in the master's program *Sprache und Kommunikation* (language and communication)
* focus: *Angewandte Kommunikations- und Sprachwissenschaft* (applied communication and language science)
* modules: MA-SK 5b, MA-SK 20, MA-MED 7/4
* when: Wednesdays, 14-16 h
* where: Phonetiklabor (phonetics lab)

# What is it about?
* automatic speech processing (*Maschinelle Sprachverarbeitung*)
* detecting speaker characteristics (e.g. emotion, age, gender) with machine learning methods
* a practical seminar based on the software [Nkululeko](https://github.com/felixbur/nkululeko/)
* a general description can be found [in my blog](http://blog.syntheticspeech.de/2021/06/22/seminar-analyze-speech-wit-respect-to-emotional-expression/)

## Professional working techniques
* besides speech processing, the seminar also teaches general techniques of professional work:
    * project management software (e.g. git)
    * data visualization with Python
    * documentation with Markdown, [LaTeX and BibTeX](http://blog.syntheticspeech.de/2022/08/25/how-to-use-latex-for-your-project-documentation/)

# What we will do in the seminar
1. create your own audio recording
    * a short talk in front of the group, or
    * a collection of YouTube data
2. segment the recording
3. annotate the segments
4. train a common model on the data of all participants

## Data donation
* we kindly ask you to donate your recordings to the institute
* so that future seminars (and research) can profit from the database

# Assessment
* [submission](https://isis.tu-berlin.de/mod/assign/view.php?id=1280229) of a seminar paper on automatic speaker classification
* written as a report in [LaTeX](https://www.overleaf.com/read/rnhsjvwxpmjh#8430b7) (Overleaf template)
* group work is possible, but each participant hands in an individual paper

## The seminar paper
* content:
    * your own understanding of machine learning (the seminar content)
    * presentation of your own results, based on the database created in the seminar
* length: 4-6 pages
* images are welcome, references to literature as well
